```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('LA_AQS_2023.csv')
```

```python
df.head()
```

```python
df_O3 = df[(df['Parameter Name']=='Ozone') & (df['Duration Description']=='1 HOUR')]
df_O3.head()
```

```python
df_O3.columns
```

```python
df_O3["Date (Local)"]
```

```python
df_O3["Arithmetic Mean"]
```

```python
date = df_O3["Date (Local)"]
value = df_O3["Arithmetic Mean"]
```

```python
fig, ax = plt.subplots(figsize=(12,6))
ax.plot(date, value)
```

```python
df_O3["Date (Local)"] = pd.to_datetime(df_O3["Date (Local)"])
```

```python
# Plot
sns.scatterplot(
  data = df_O3,
  x = df_O3["Date (Local)"],
  y = df_O3["Arithmetic Mean"]
)
plt.show()
```
